```python
import matplotlib.pyplot as plt
import numpy as np

print("=== DEMONSTRATING MATPLOTLIB PLOTS ===")

# Set random seed for reproducible scatter and histogram data
np.random.seed(42)

# ==========================================
# 1. LINE PLOT (plt.plot)
# ==========================================
plt.figure(figsize=(8, 5))
x_line = [1, 2, 3, 4, 5]
y_line = [2, 4, 6, 8, 10]

plt.plot(x_line, y_line, color='blue', marker='o', linestyle='-', linewidth=2)
plt.title("1. Line Plot Example")
plt.xlabel("X-Axis (Time / Steps)")
plt.ylabel("Y-Axis (Growth Value)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


# ==========================================
# 2. BAR PLOT (plt.bar)
# ==========================================
plt.figure(figsize=(8, 5))
categories = ['Electronics', 'Clothing', 'Groceries', 'Books']
sales = [45000, 30000, 25000, 15000]

plt.bar(categories, sales, color='orange', edgecolor='black')
plt.title("2. Bar Plot Example (Category Comparison)")
plt.xlabel("Product Categories")
plt.ylabel("Total Sales ($)")
plt.show()


# ==========================================
# 3. SCATTER PLOT (plt.scatter)
# ==========================================
plt.figure(figsize=(8, 5))
x_scatter = np.random.rand(50) * 100
y_scatter = np.random.rand(50) * 100

plt.scatter(x_scatter, y_scatter, color='green', alpha=0.7, s=80, edgecolors='black')
plt.title("3. Scatter Plot Example (Correlation Check)")
plt.xlabel("Feature A")
plt.ylabel("Feature B")
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()


# ==========================================
# 4. HISTOGRAM (plt.hist)
# ==========================================
plt.figure(figsize=(8, 5))
# Generate 1,000 random values from a normal distribution
data_hist = np.random.randn(1000)

plt.hist(data_hist, bins=30, color='purple', edgecolor='black', alpha=0.75)
plt.title("4. Histogram Example (Distribution Frequency)")
plt.xlabel("Value Ranges")
plt.ylabel("Frequency")
plt.show()
```
