```python
import math

print("=== DEMONSTRATING PYTHON DATA TYPES AND OPERATIONS ===")
print("=" * 55)

# ==========================================
# 1. INTEGER (int)
# ==========================================
print("\n--- 1. Integer (int) ---")
a = 15
b = 4
print(f"Variables: a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")
print(f"Floor Division (a // b): {a // b}")
print(f"Modulo / Remainder (a % b): {a % b}")
print(f"Built-in abs(-15): {abs(-15)}")
print(f"Built-in pow(2, 3): {pow(2, 3)}")
print(f"Math module - factorial(5): {math.factorial(5)}")
print(f"Math module - gcd(24, 36): {math.gcd(24, 36)}")
print(f"Math module - isqrt(25): {math.isqrt(25)}")

# ==========================================
# 2. FLOAT (float)
# ==========================================
print("\n--- 2. Float (float) ---")
x = 10.75
y = 3.2
print(f"Variables: x = {x}, y = {y}")
print(f"Multiplication (x * y): {x * y}")
print(f"Division (x / y): {x / y}")
print(f"Built-in round(x, 1): {round(x, 1)}")
print(f"Math module - ceil(x): {math.ceil(x)}")
print(f"Math module - floor(x): {math.floor(x)}")
print(f"Math module - sqrt(16.0): {math.sqrt(16.0)}")

# ==========================================
# 3. COMPLEX (complex)
# ==========================================
print("\n--- 3. Complex (complex) ---")
c1 = 3 + 4j
c2 = 1 - 2j
print(f"Variables: c1 = {c1}, c2 = {c2}")
print(f"Addition (c1 + c2): {c1 + c2}")
print(f"Real part of c1: {c1.real}")
print(f"Imaginary part of c1: {c1.imag}")
print(f"Conjugate of c1: {c1.conjugate()}")
print(f"Absolute value / Magnitude abs(c1): {abs(c1)}")

# ==========================================
# 4. STRING (str)
# ==========================================
print("\n--- 4. String (str) ---")
s = "Hello, Python World!"
print(f"String: '{s}'")
print(f"Length (len): {len(s)}")
print(f"Uppercase: {s.upper()}")
print(f"Lowercase: {s.lower()}")
print(f"Slicing s[0:5]: {s[0:5]}")
print(f"Replacement: {s.replace('World', 'Universe')}")
print(f"Split by comma: {s.split(',')}")

# ==========================================
# 5. LIST (list - Mutable Sequence)
# ==========================================
print("\n--- 5. List (list) ---")
fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")
fruits.append("orange")
print(f"After append('orange'): {fruits}")
fruits.insert(1, "blueberry")
print(f"After insert(1, 'blueberry'): {fruits}")
fruits.sort()
print(f"After sort(): {fruits}")
print(f"Total items (len): {len(fruits)}")

# ==========================================
# 6. TUPLE (tuple - Immutable Sequence)
# ==========================================
print("\n--- 6. Tuple (tuple) ---")
coordinates = (10.5, 20.3, 30.1, 20.3)
print(f"Tuple: {coordinates}")
print(f"First element coordinates[0]: {coordinates[0]}")
print(f"Count of 20.3: {coordinates.count(20.3)}")
print(f"Index of 30.1: {coordinates.index(30.1)}")

# ==========================================
# 7. DICTIONARY (dict - Key-Value Pairs)
# ==========================================
print("\n--- 7. Dictionary (dict) ---")
student = {"name": "Alice", "age": 22, "course": "Data Science"}
print(f"Dictionary: {student}")
print(f"Accessing 'name': {student['name']}")
print(f"Keys: {list(student.keys())}")
print(f"Values: {list(student.values())}")
student.update({"age": 23})
print(f"After updating age: {student}")

# ==========================================
# 8. SET (set - Unordered Unique Collection)
# ==========================================
print("\n--- 8. Set (set) ---")
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(f"set1: {set1}, set2: {set2}")
print(f"Union (set1 | set2): {set1 | set2}")
print(f"Intersection (set1 & set2): {set1 & set2}")
print(f"Difference (set1 - set2): {set1 - set2}")
set1.add(6)
print(f"After adding 6 to set1: {set1}")

# ==========================================
# 9. FROZENSET (frozenset - Immutable Set)
# ==========================================
print("\n--- 9. Frozenset (frozenset) ---")
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4, 5])
print(f"Frozenset 1: {fs1}")
print(f"Frozenset 2: {fs2}")
print(f"Intersection (fs1 & fs2): {fs1 & fs2}")
print(f"Union (fs1 | fs2): {fs1 | fs2}")

# ==========================================
# 10. BOOLEAN (bool)
# ==========================================
print("\n--- 10. Boolean (bool) ---")
is_active = True
is_admin = False
print(f"is_active: {is_active}, is_admin: {is_admin}")
print(f"Logical AND (is_active and is_admin): {is_active and is_admin}")
print(f"Logical OR (is_active or is_admin): {is_active or is_admin}")
print(f"Logical NOT (not is_admin): {not is_admin}")

# ==========================================
# 11. NONETYPE (None)
# ==========================================
print("\n--- 11. NoneType (None) ---")
result = None
print(f"Value of result: {result}")
print(f"Type of result: {type(result)}")
if result is None:
    print("Conditional check: The variable 'result' is currently None (uninitialized).")
```
