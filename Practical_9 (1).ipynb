```python
import matplotlib.pyplot as plt
import seaborn as sns

print("=== STATISTICAL PLOTS USING SEABORN ===")

# Set Seaborn visual theme for aesthetic styling
sns.set_theme(style="whitegrid")

# Load sample datasets provided directly by Seaborn
tips_df = sns.load_dataset("tips")
flights_df = sns.load_dataset("flights")

print(f"Tips dataset preview:\n{tips_df.head(3)}\n")

# ==========================================
# 1. SCATTER PLOT (sns.scatterplot)
# ==========================================
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=tips_df,
    x="total_bill",
    y="tip",
    hue="day",
    style="time",
    palette="deep",
    s=80,
)
plt.title("1. Scatter Plot: Total Bill vs. Tip (Categorized by Day & Time)")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()


# ==========================================
# 2. LINE PLOT (sns.lineplot)
# ==========================================
plt.figure(figsize=(9, 5))
# Showing trends over time using the flights dataset
sns.lineplot(
    data=flights_df, x="year", y="passengers", hue="month", palette="tab10"
)
plt.title("2. Line Plot: Monthly Passenger Trends Across Years")
plt.xlabel("Year")
plt.ylabel("Passenger Count")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left", title="Month")
plt.tight_layout()
plt.show()


# ==========================================
# 3. BAR PLOT (sns.barplot)
# ==========================================
plt.figure(figsize=(8, 5))
# Shows means and confidence intervals automatically
sns.barplot(
    data=tips_df, x="day", y="total_bill", hue="sex", palette="muted", ci=None
)
plt.title("3. Bar Plot: Average Total Bill by Day and Sex")
plt.xlabel("Day of the Week")
plt.ylabel("Average Total Bill ($)")
plt.show()


# ==========================================
# 4. HISTOGRAM WITH KDE (sns.histplot)
# ==========================================
plt.figure(figsize=(8, 5))
# kde=True overlays a smooth Kernel Density Estimate curve
sns.histplot(
    data=tips_df, x="total_bill", kde=True, color="purple", bins=20, alpha=0.6
)
plt.title("4. Histogram & KDE: Distribution of Total Bills")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency / Density")
plt.show()


# ==========================================
# 5. BOX PLOT (sns.boxplot)
# ==========================================
plt.figure(figsize=(8, 5))
# Shows medians, quartiles, and outliers across categories
sns.boxplot(
    data=tips_df, x="day", y="total_bill", hue="smoker", palette="Set2"
)
plt.title("5. Box Plot: Total Bill Distribution by Day and Smoker Status")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")
plt.show()
```
