```python
import numpy as np

print("=== RANDOM NUMBER GENERATION & STATISTICAL SIMULATION ===")
print("=" * 60)

# Set a random seed for reproducibility (ensures you get the same random numbers each time)
np.random.seed(42)

# ==========================================
# 1. np.random.rand (Uniform Distribution)
# ==========================================
print("\n--- 1. Uniform Distribution [0, 1) using np.random.rand ---")
# Generates random floats uniformly distributed over [0, 1)
# Shape: 3 rows, 2 columns
uniform_arr = np.random.rand(3,  2)
print(f"Uniform random array:\n{uniform_arr}")


# ==========================================
# 2. np.random.randn (Standard Normal Distribution)
# ==========================================
print("\n--- 2. Standard Normal Distribution using np.random.randn ---")
# Generates samples from a standard normal distribution (Mean = 0, Standard Deviation = 1)
# Shape: 1D array with 5 elements
standard_normal_arr = np.random.randn(5)
print(f"Standard normal samples: {standard_normal_arr}")


# ==========================================
# 3. np.random.randint (Random Integers)
# ==========================================
print("\n--- 3. Random Integers using np.random.randint ---")
# Generates random integers from low (inclusive) to high (exclusive)
# Syntax: np.random.randint(low, high, size)
random_ints = np.random.randint(1, 10, size=(2, 4))
print(f"Random integers between 1 and 9 (2x4 matrix):\n{random_ints}")


# ==========================================
# 4. np.random.normal & COMPUTING STATISTICS
# ==========================================
print("\n--- 4. Custom Normal Distribution & Statistics (np.random.normal) ---")
# Simulating a dataset (e.g., student heights):
# Mean (loc) = 170 cm, Standard Deviation (scale) = 10 cm, Sample Size = 1000
simulated_heights = np.random.normal(loc=170, scale=10, size=1000)

print(f"First 5 simulated heights: {simulated_heights[:5]}")

# Computing descriptive statistics on the simulated distribution
print(f"\nComputed Statistics:")
print(f"Mean: {np.mean(simulated_heights):.2f} cm")
print(f"Standard Deviation: {np.std(simulated_heights):.2f} cm")
print(f"Minimum Value: {np.min(simulated_heights):.2f} cm")
print(f"Maximum Value: {np.max(simulated_heights):.2f} cm")
print(f"Median: {np.median(simulated_heights):.2f} cm")
```
