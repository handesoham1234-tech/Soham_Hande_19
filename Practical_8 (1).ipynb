```python
import matplotlib.pyplot as plt
import numpy as np

print("=== DEMONSTRATING SUBPLOTS, STYLING, LEGENDS, & GRIDS ===")

# Generate sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = x
y4 = x ** 2

# Create a figure with a custom size
plt.figure(figsize=(12, 10))

# ==========================================
# 1. FIRST SUBPLOT (Top Panel: 2x1 Grid)
# ==========================================
# plt.subplot(nrows, ncols, index_or_plot_number)
plt.subplot(2, 1, 1)

# Plotting multiple lines with styling (color, linestyle, linewidth)
plt.plot(x, y1, color='crimson', linestyle='-', linewidth=2.5, label='Sine Wave')
plt.plot(x, y2, color='dodgerblue', linestyle='--', linewidth=2, label='Cosine Wave')

# Using the requested functions for Subplot 1
plt.title("Trigonometric Functions (Sine & Cosine)")
plt.xlabel("X Values")
plt.ylabel("Amplitude")
plt.legend(loc='upper right', frameon=True)
plt.grid(True, linestyle=':', alpha=0.7)


# ==========================================
# 2. SECOND SUBPLOT (Bottom Panel: 2x1 Grid)
# ==========================================
plt.subplot(2, 1, 2)

# Plotting polynomial and linear data with styling
plt.plot(x, y3, color='forestgreen', linestyle='-.', linewidth=2, label='Linear (y = x)')
plt.plot(x, y4, color='darkorange', linestyle='-', linewidth=2.5, label='Quadratic (y = x^2)')

# Using the requested functions for Subplot 2
plt.title("Growth Functions (Linear vs. Quadratic)")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend(loc='upper left', frameon=True)
plt.grid(True, linestyle='--', alpha=0.7)


# Adjust layout to prevent overlapping titles and labels
plt.tight_layout()

# Display the multi-plot figure
plt.show()
```
