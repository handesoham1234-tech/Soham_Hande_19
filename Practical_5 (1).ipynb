```python
import pandas as pd

print("=== DEMONSTRATING PANDAS SERIES, DATAFRAMES, & CSV OPERATIONS ===")
print("=" * 65)

# ==========================================
# 1. CREATING A PANDAS SERIES
# ==========================================
print("\n--- 1. Pandas Series (1D Labeled Array) ---")
scores = [85, 90, 78, 92, 88]
labels = ["Math", "Science", "English", "History", "Computer"]

# Using pd.Series with custom index labels
series_data = pd.Series(scores, index=labels)
print(f"Series Object:\n{series_data}")
print(f"Mean Score: {series_data.mean()}")


# ==========================================
# 2. CREATING A PANDAS DATAFRAME
# ==========================================
print("\n--- 2. Pandas DataFrame (2D Tabular Data) ---")

data = {
    "Employee": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Department": ["IT", "HR", "Finance", "IT", "Marketing"],
    "Age": [28, 34, 41, 29, 36],
    "Salary": [75000, 62000, 88000, 72000, 68000]
}

# Using pd.DataFrame
df = pd.DataFrame(data)
print(f"Original DataFrame:\n{df}")


# ==========================================
# 3. EXPORTING TO A CSV FILE (df.to_csv)
# ==========================================
print("\n--- 3. Exporting to CSV ---")
csv_filename = "employees.csv"
df.to_csv(csv_filename, index=False)
print(f"DataFrame successfully exported to '{csv_filename}'")


# ==========================================
# 4. IMPORTING FROM A CSV FILE (pd.read_csv)
# ==========================================
print("\n--- 4. Importing from CSV ---")
df_imported = pd.read_csv(csv_filename)
print(f"Imported DataFrame from '{csv_filename}':\n{df_imported}")


# ==========================================
# 5. BASIC DATAFRAME OPERATIONS & INSPECTION
# ==========================================
print("\n--- 5. Basic DataFrame Operations ---")

# df.head() - View first few rows (default is 5, let's look at 3)
print(f"\nUsing df.head(3):\n{df.head(3)}")

# df.tail() - View last few rows (default is 5, let's look at 2)
print(f"\nUsing df.tail(2):\n{df.tail(2)}")

# df.describe() - Statistical summary of numerical columns
print(f"\nUsing df.describe():\n{df.describe()}")

# Additional inspection attributes
print(f"\nDataFrame Shape (Rows, Columns): {df.shape}")
print(f"DataFrame Columns: {list(df.columns)}")
```
