```python
import pandas as pd
import numpy as np

# 1. Create a sample DataFrame with missing values
raw_data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Department': ['HR', 'IT', 'IT', 'HR', 'Sales', 'Sales'],
    'Salary': [50000, 75000, np.nan, 48000, 60000, np.nan],
    'Bonus': [5000, np.nan, 3000, 4000, np.nan, np.nan]
}
df = pd.DataFrame(raw_data)

# 2. Handle Missing Values
print("--- Missing Value Count ---")
print(df.isnull().sum())

# Fill missing Salaries with the column mean, and drop rows missing vital Bonus markers
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df_cleaned = df.dropna(subset=['Bonus']) 

print("\n--- Cleaned Data (Nulls handled) ---")
print(df_cleaned)

# 3. Filter Rows and Columns
# Keep rows where Salary > 49000, select only 'Employee' and 'Salary' columns
filtered_df = df_cleaned[df_cleaned['Salary'] > 49000][['Employee', 'Salary']]

# 4. Sort Values
# Sort employees by Salary in descending order
sorted_df = df_cleaned.sort_values(by='Salary', ascending=False)
print("\n--- Sorted by Salary (Descending) ---")
print(sorted_df)

# 5. Group and Aggregate
# Group by Department and calculate the average Salary and total Bonus
summary_df = df_cleaned.groupby('Department').agg({
    'Salary': 'mean',
    'Bonus': 'sum'
}).reset_index()

print("\n--- Department Performance Summary ---")
print(summary_df)
```
