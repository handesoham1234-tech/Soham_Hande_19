```python
import numpy as np

print("=== DEMONSTRATING NUMPY ARRAYS & OPERATIONS ===")
print("=" * 50)

# ==========================================
# 1. CREATING 1D ARRAYS
# ==========================================
print("\n--- 1. Creating 1D Arrays ---")

# Using np.array
arr_1d = np.array([10, 20, 30, 40, 50])
print(f"np.array: {arr_1d} | Type: {type(arr_1d)}")

# Using np.arange (start, stop, step)
arr_range = np.arange(1, 10, 2)
print(f"np.arange (1 to 10, step 2): {arr_range}")

# Using np.linspace (start, stop, num_elements)
arr_linspace = np.linspace(0, 1, 5)
print(f"np.linspace (0 to 1, 5 points): {arr_linspace}")


# ==========================================
# 2. CREATING & RESHAPING 2D ARRAYS
# ==========================================
print("\n--- 2. Creating & Reshaping 2D Arrays ---")

# Create a 1D sequence from 1 to 12
flat_arr = np.arange(1, 13)
print(f"Original flat array (12 elements): {flat_arr}")

# Using np.reshape to convert 1D into a 3x4 2D matrix
matrix_2d = np.reshape(flat_arr, (3, 4))
print(f"Reshaped 2D Array (3 rows, 4 columns):\n{matrix_2d}")


# ==========================================
# 3. INDEXING AND SLICING
# ==========================================
print("\n--- 3. Indexing and Slicing ---")

# 1D Indexing & Slicing
print(f"1D Element at index 2: {arr_1d[2]}")
print(f"1D Slices from index 1 to 3: {arr_1d[1:4]}")

# 2D Indexing [row, column]
print(f"2D Element at row 1, col 2: {matrix_2d[1, 2]}")

# 2D Slicing (Rows 0 to 1, Columns 1 to 3)
print(f"2D Sub-matrix slice (Rows 0-1, Cols 1-3):\n{matrix_2d[0:2, 1:4]}")


# ==========================================
# 4. MATHEMATICAL FUNCTIONS (np.sum, np.mean)
# ==========================================
print("\n--- 4. Mathematical Functions ---")

print(f"Total Sum of matrix_2d: {np.sum(matrix_2d)}")
print(f"Overall Mean of matrix_2d: {np.mean(matrix_2d):.2f}")

# Operations across specific axes
# axis=0 operates vertically (down columns)
print(f"Sum along columns (axis=0): {np.sum(matrix_2d, axis=0)}")

# axis=1 operates horizontally (across rows)
print(f"Mean along rows (axis=1): {np.mean(matrix_2d, axis=1)}")
```
