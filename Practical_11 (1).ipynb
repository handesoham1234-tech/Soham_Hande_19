```python
import numpy as np

print("=== LINEAR ALGEBRA OPERATIONS & MATRIX EQUATIONS ===")
print("=" * 55)

# Define a system of linear equations:
# 2x + 1y = 5
# 3x + 4y = 10
# Represented as Matrix A (coefficients) and Vector b (constants)
A = np.array([[2, 1], 
              [3, 4]])
b = np.array([5, 10])

print(f"Coefficient Matrix A:\n{A}")
print(f"Constants Vector b: {b}")


# ==========================================
# 1. MATRIX TRANSPOSE (.T)
# ==========================================
print("\n--- 1. Matrix Transpose ---")
# Using a 2x3 rectangular matrix to clearly demonstrate row/column flipping
rect_matrix = np.array([[1, 2, 3], 
                        [4, 5, 6]])
transposed_matrix = rect_matrix.T

print(f"Original Rectangular Matrix (Shape {rect_matrix.shape}):\n{rect_matrix}")
print(f"Transposed Matrix (Shape {transposed_matrix.shape}):\n{transposed_matrix}")


# ==========================================
# 2. MATRIX INVERSE (np.linalg.inv)
# ==========================================
print("\n--- 2. Matrix Inverse ---")
# Computes the multiplicative inverse of a square matrix (A^-1)
A_inv = np.linalg.inv(A)
print(f"Inverse of Matrix A (A^-1):\n{A_inv}")

# Verification: Multiplying a matrix by its inverse yields the Identity matrix
identity_check = np.dot(A, A_inv)
print(f"Verification (A * A_inv -> Rounded Identity):\n{np.round(identity_check)}")


# ==========================================
# 3. SOLVING MATRIX EQUATIONS (np.linalg.solve)
# ==========================================
print("\n--- 3. Solving Matrix Equations (Ax = b) ---")
# Efficiently solves the linear system Ax = b for x without computing full inverses
solution = np.linalg.solve(A, b)
print(f"Solution vector x (x, y): {solution}")
print(f"Interpretation: x = {solution[0]:.2f}, y = {solution[1]:.2f}")


# ==========================================
# 4. EIGENVALUES & EIGENVECTORS (np.linalg.eig)
# ==========================================
print("\n--- 4. Eigenvalues and Eigenvectors ---")
# Computes the eigenvalues and right eigenvectors of a square array
eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors Matrix:\n{eigenvectors}")
```
